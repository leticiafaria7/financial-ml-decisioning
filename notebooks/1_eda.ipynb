## 2. Distribuições bivariadas

### Como é a distribuição de idade por emprego?

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde
from matplotlib.colors import LinearSegmentedColormap
import seaborn as sns

# Configurações
plt.rcParams["font.family"] = "DejaVu Sans"

dados = df1[["job", "age"]].dropna().copy()

# Ordena os jobs pela mediana da idade
ordem = dados.groupby("job")["age"].median().sort_values().index

# Paleta verde → laranja → marrom
cmap = LinearSegmentedColormap.from_list(
    "verde_laranja",
    ["#1B7F5C", "#42A66F", "#8EBF72", "#D5B65A", "#D98245", "#A65335", "#6F3B2C"]
)

cores = [cmap(i / max(len(ordem) - 1, 1)) for i in range(len(ordem))]

# Eixo X de 0 a 100
x = np.linspace(0, 100, 500)

fig, ax = plt.subplots(figsize=(10, 8))

fig.patch.set_facecolor("#FAF8F3")
ax.set_facecolor("#FAF8F3")

for i, (job, cor) in enumerate(zip(ordem, cores)):
    idade = dados.loc[dados["job"] == job, "age"]

    if len(idade) < 2 or idade.nunique() < 2:
        continue

    kde = gaussian_kde(idade)
    kde.set_bandwidth(kde.factor * 1.3)

    densidade = kde(x)
    densidade /= densidade.max()

    y_base = i
    altura = 2.0
    y = y_base + densidade * altura

    ax.fill_between(x, y_base, y, color=cor, alpha=0.40)
    ax.plot(x, y, color=cor, linewidth=2)
    ax.hlines(y_base, 0, 100, color=cor, linewidth=1, alpha=0.55)

# Eixos
ax.set_yticks(range(len(ordem)))
ax.set_yticklabels(ordem, fontsize=11)
ax.set_xlim(0, 100)
ax.set_xlabel("Idade", fontsize=12, color="#4A403A")
ax.set_ylabel("")
ax.set_title("Distribuição de idade por profissão", loc="center", fontsize=14, fontweight="normal", color="#4A403A", pad=20)
ax.tick_params(axis="y", length=0)
ax.tick_params(axis="both", colors="#4A403A")

sns.despine(ax=ax, left=True, top=True, right=True)

plt.tight_layout()
plt.show()
```

### Quando uma variável é unknown, as outras também são?

```python
tb_unknown = pd.DataFrame()

for c in df1.columns:
    tmp = df1[df1[c] == 'unknown']
    if tmp.shape[0] == 0:
        continue
    else:
        df_tmp = pd.DataFrame(data = {'coluna':[c], 'volume':[tmp.shape[0]], 'pct':[round(tmp.shape[0] * 100 / df1.shape[0], 2)]})

        tb_unknown = pd.concat([tb_unknown, df_tmp])

print(f"Variáveis com a categoria 'unknown'\n{'-' * 50}")
tb_unknown
```
