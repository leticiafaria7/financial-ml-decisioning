## Teste da API em produção

### Imports

```python
import requests

url = "https://financial-ml-decisioning.onrender.com/"
```

### GET /api/categories

```python
categories = f"{url}/api/categories"
resp = requests.get(categories).json()

resp
```

### GET /api/metrics

```python
metrics = f"{url}/api/metrics"
resp = requests.get(metrics).json()

resp
```

### GET /api/indicators

```python
params = {'month':'05. may', 'year':2008, 'month_position':'início'}

indicators = f"{url}/api/indicators"
resp = requests.get(indicators, params = params).json()

resp
```

### POST /api/predict

```python
payload = {
  "age": 35,
  "day_of_week": "1. mon",
  "education": "university.degree",
  "housing": "unknown",
  "job": "admin.",
  "loan": "unknown",
  "marital": "married",
  "month": "05. may",
  "month_position": "meio",
  "poutcome": "nonexistent",
  "previous": 0,
  "year": 2009
}

resp = requests.post(f"{url}/api/predict", json = payload)
print(resp.json())
```
